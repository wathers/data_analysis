```python
import pandas as pd
pd.__version__
```

```python
df = pd.read_csv('../employees_salary.csv')
g1 = df.groupby('部门') #查看分组
g1.get_group("人事部") # 查看详情
g1[['工资(元)']].mean().round(2)
```

```python
## 多个列进行分组
g2 = df.groupby(['姓名','部门'])
g2[['工资(元)']].mean().sort_values("工资(元)",ascending=False)
```
