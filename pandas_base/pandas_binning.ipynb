```python
import pandas as pd
```

```python
df = pd.read_csv("../employees_salary.csv")
df
```

```python
# 分箱
pd.cut(df['工资(元)'], bins=2)
print(pd.cut(df['工资(元)'], bins=2).value_counts())
```

```python
# 自定义区间
df['工资区间'] = pd.cut(df['工资(元)'], bins=[3000, 10000, 20000, 30000], labels=['低', '中', '高'])
df['工资区间']
```

```python
# 区间等频率的分割
pd.qcut(df['工资(元)'], q=3).value_counts()
```

```python
df.reset_index(inplace=True)
df.set_index('员工编号',inplace=True)
df
```
