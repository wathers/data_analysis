```python
import pandas as pd
```

### 数据的变形

```python
data = {
    'ID':[1,2],
    'Name':['Alice','Bob'],
    'Math':[80,90],
    'English':[74,80],
    'Science':[85,92],
    'History':[78,88],
    'Geography':[75,85]
}
df = pd.DataFrame(data)
df
```

```python
#转置
df.T
```

```python
#宽表-->长表
df2 = df.melt(id_vars=['ID','Name'],value_vars=['English','Science',"History","Geography"],var_name="科目",value_name="分数").sort_values('ID')
df2
```

```python
#长表-->宽表
df2.pivot(index=['ID',"Name"],columns=["科目"],values="分数")
```

```python
#分列，把Name列中的值 取首字母作为一列，最后一个字母作为一列
df['First_Letter'] = df['Name'].str[0]
df['Last_Letter'] = df['Name']
df.info()
```
