```python
import pandas as pd
```

```python
data = {
    'name': ['Alice', 'Alice', 'Bob', 'Charlie', 'Alice', 'Bob'],
    'age': [25, 26, 30, 35, 25, 21],
    'male': ['Male', 'Female', 'Female', 'Female', 'Male', 'Male'],
    'city': ['New York', 'New York', 'Los Angeles', 'Chicago', 'New York', 'Los Angeles']
}
df = pd.DataFrame(data)
print(df)
#检查重复
df.duplicated()
#根据name去重
df.drop_duplicates(subset=['name'])
#留最新的数据 根据index判断 keep 默认是first
df.drop_duplicates(subset=['name'], keep='last')
```

```python
## 数据类型转换
## 使用astype
df['age'] = df['age'].astype('int8')
df['male'] = df['male'].astype('category')
df.dtypes
df['sex'] = df['male'].map({'Male': 0, 'Female': 1})
df.dtypes
```
