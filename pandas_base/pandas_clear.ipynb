```python
import pandas as pd
```

### 数据的缺失值的删除

```python
df = pd.read_csv('../mock_data_missing.csv')
#查看是否为缺失值
df.isna()
#查看缺失值的总和
df.isna().sum()
#删除缺失值
print(f"默认参数:{ df.dropna().index.size }")
print(f"how='all':{ df.dropna(how='all').index.size }")
print(f"至少有n个不是缺失值就保留:{ df.dropna(thresh=10).index.size }")
print(f"删除一整列:\n",df.dropna(axis=1))
print(f"如果指定列有缺失值就删除这一行:\n{df.dropna(subset=['date'])}")
```

### 缺失值的填充

```python
df = pd.read_csv('../mock_data_missing.csv')
#固定值填充
df.fillna({'date':'2020-01-01','region':'华中'})
#使用统计值来填充
df.fillna({'quantity':df.quantity.mean()})
#根据缺失值附近的值填充
df.ffill()
df.bfill()
```
