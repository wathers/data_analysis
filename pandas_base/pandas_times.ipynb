```python
# 时间间隔
df = pd.DataFrame({
    'data1':['2025-01-01 23:13:00', '2025-01-02 23:13:00', '2025-01-03 23:13:00'] ,
    'data2':['2026-01-01 23:13:00', '2026-01-02 23:13:00', '2026-01-03 23:13:00']
})
df['datetime1'] = pd.to_datetime(df['data1'])
df['datetime2'] = pd.to_datetime(df['data2'])
df['delta'] = df['datetime2'] - df['datetime1']
df
```
