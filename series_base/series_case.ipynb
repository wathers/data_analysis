## 创建一个学生成绩 50-100分之间，计算最高分、最低分、平均分，高于平均分的人数

```python
np.random.seed(1)
arr = np.random.randint(50,100,30)
scores = pd.Series(arr)
print(f"平均分数,{scores.mean()}")
print(f"最高分,{scores.max()}")
print(f"最低分,{scores.min()}")
print(f"高于评价分数的人数,{scores[scores>scores.mean()].count()}")
scores[scores>scores.mean()].count()
```

### 一周的天气 Series中
1. 找出超过30℃的日期
2. 计算平均温度
3. 温度从高到低排序
4. 找出温度变化最大的两天

```python
temperature = pd.Series([30,32,38,28,29,27,26],name="温度",index=["周一","周二","周三","周四","周五","周六","周日"])
print(f"超过30℃的日期,{temperature[temperature>30].index}")
print(f"平均温度,{temperature.mean()}")
print(f"温度从高到低排序,{temperature.sort_values(ascending=False)}")
print(f"温度变化最大的两天,{temperature.diff().idxmax()},{temperature.diff().idxmin()}")
```
