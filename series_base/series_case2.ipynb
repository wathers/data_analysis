### 24小时营业额分析
1. 按天查询24的营业额
2. 查询8:00-22:00的营业额和非工作时间的营业额
3. 查询最高的3小时营业额

```python
np.random.seed(1)
ddata = np.random.randint(100,200,24)
ddata
```

```python
arr = pd.Series(ddata,index=pd.date_range('2026-01-01 00:00:00',freq='h',periods=24),name='营业额')
```

```python
oneday = arr.resample(rule='D').sum()
print(f'一天的营业额为：{oneday.values[0]}')
```

```python
yingye = arr[(arr.index.hour>=8) & (arr.index.hour<=22)].sum()
nonyingye = arr[(arr.index.hour<8) | (arr.index.hour>22)].sum()
print(f"营业时间/非营业时间: {yingye/nonyingye:0.2f}")
#arr.between_time('8:00',"22:00") #按时间来查询
```
