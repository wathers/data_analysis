```python
import pandas as pd
```

### 场景:某公司销售数据如下，请完成以下任务:
1.计算每种产品[总销售额(销售额=单价销量)。
2.找出销售额最合的产品。
3.按销售额从高到低排序，并输出所有产品信息。

```python
data = {
    '产品名称': ['A', 'B', 'C', 'D'],
    '单价': [100, 150, 200, 120],
    '销量': [50, 30, 20, 40]
}

df=pd.DataFrame(data)
df
```

```python
# 计算每种产品[总销售额(销售额=单价销量)。
df['总销售额'] = df['单价'] * df['销量']
df
```

```python
#找出销售额最合的产品。
df.nlargest(1,'总销售额')
```

```python
#按销售额从高到低排序，并输出所有产品信息。
df.sort_values(['总销售额'],ascending=False)
```
