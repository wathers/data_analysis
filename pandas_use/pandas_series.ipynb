```python
import pandas as pd
```

## series

### series的创建

```python
#直接创建
arr = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'],dtype='int8',name='arr')
print(arr)
```

```python
#通过字典来创建
dd = {'a':1,'b':2,'c':3,'d':4,'e':5}
arr = pd.Series(dd,index=['a','b'])
print(arr)
```

### series的属性

```python
#隐式索引
print(f"维度,{arr.ndim}")
print(f"形状,{arr.shape}")
print(f"dtype,{arr.dtype}")
print(f"size,{arr.size}")
print(f"获取index对象,{arr.index}")
print(f"获取value对象,{arr.values}")
```

### series获取数据

```python
print(f"显式索引获取 arr['a'],{arr['a']}")
print(f"显式索引获取 arr.loc['a'],{arr.loc['a']}")
print(f"隐式索引获取 arr[0],{arr.iloc[0]}")
print(f"显式索引切片 arr.loc['a':],{arr.loc['a':]}")
print(f"隐式索引切片 arr.iloc[1:],{arr.iloc[1:]}")
print(f"at的方式获取 arr.at['a'],{arr.at['a']}")
print(f"at的方式获取 arr.iat[0],{arr.iat[0]}")
print(f"bool筛选 arr[arr<1],{arr[arr>1]}")
```

### series常用方法

```python
print(f"获取前几个 arr.head(1)，{arr.head(1)}")
print(f"获取后几个 arr.tail(1)，{arr.tail(1)}")
print(f"判断元素是否在集合里面 isin(),{arr.isin([6,7])}")
print(f"判断元素是否为NaN, {arr.isna()}")
print(f"获取元素的个数,{arr.count()}")
print("获取统计信息")
arr.describe()
```
