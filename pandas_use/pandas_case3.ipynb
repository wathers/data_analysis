```python
import pandas as pd
from pandas.core._numba.kernels import mean_
```

```python
# 2.导入数据
df = pd.read_csv('../penguins_data.csv')
# 3.数据的清洗
df.isna().sum()
df = df.dropna()
df['性别'] = df['性别'].astype('category')
# 4.数据特征构件
df['喙长_喙深比'] = df['喙长_mm'] / df['喙深_mm']
# 5.数据分析
df['体重分类'] = pd.cut(df['体重_g'],bins=3,labels=['低','中','高'])
## 按物种分组
g1 = df.groupby(["物种"])
g1.agg({
    '体重_g':['mean','sum']
})
g2 = df.groupby(['物种','性别']).agg({
    '体重_g':['mean']
})
g2
```
