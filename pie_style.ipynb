```python
import matplotlib.pyplot as plt
from matplotlib import colors
from wcwidth import width

plt.rcParams['font.sans-serif'] = ['PingFang SC']
# 2. 解决负号 '-' 显示为方块的问题
plt.rcParams['axes.unicode_minus'] = False
```

```python
datas = ['苹果', '香蕉', '橘子', '葡萄']
values = [10, 20, 30, 40]

plt.pie(values, labels=datas,
        autopct='%.1f%%',
        startangle=0,
        colors=['red', 'yellow', 'orange', 'purple'],
        )
plt.show()
```

```python
datas = ['苹果', '香蕉', '橘子', '葡萄']
values = [10, 20, 30, 40]

plt.pie(values, labels=datas,
        autopct="%.1f%%",
        wedgeprops={"width": 0.4},
        pctdistance=0.8)
plt.text(0,0,"水果占比",ha='center',va='center')
```

```python
datas = ['苹果', '香蕉', '橘子', '葡萄']
values = [10, 20, 30, 40]

explode = (0, 0.2, 0, 0)

plt.pie(values, labels=datas,
        autopct='%.1f%%',
        startangle=0,
        explode=explode,
        )

plt.show()
```
