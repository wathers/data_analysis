```python
import random

import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['PingFang SC']
# 2. 解决负号 '-' 显示为方块的问题
plt.rcParams['axes.unicode_minus'] = False
```

```python
values = []
data = []
for i in range(1000):
    x = random.uniform(1.0, 100.0)
    data.append(x)
    values.append(x * random.gauss(1, 0.1))

plt.scatter(values, data,s=10,alpha=0.8)
plt.plot([0,120],[0,100],color='red')
plt.grid(True)

plt.show()
```
