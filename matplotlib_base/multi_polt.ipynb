```python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['PingFang SC']
# 2. 解决负号 '-' 显示为方块的问题
plt.rcParams['axes.unicode_minus'] = False
```

```python
data = {
    '语文': [80, 75, 90, 74, 85, 60, 75, 80, 90, 74, 85, 60],
    '数学': [85, 90, 75, 80, 95, 70, 85, 90, 75, 80, 95, 70],
    '英语': [65, 70, 85, 60, 75, 80, 90, 74, 85, 60, 75, 80]
}
plt.subplot(3,2,1)
plt.boxplot(data.values(), tick_labels=data.keys())
plt.title('成绩分布')
plt.xlabel('科目')
plt.grid(True,axis='y')

plt.subplot(3,2,2)
plt.hist(data['语文'], bins=10, edgecolor='black')

plt.subplot(3,2,3)
plt.barh(
    data['数学'],
    data['语文'],
    color='blue'
)
plt.subplot(2,2,4)
plt.barh(
    data['数学'],
    data['语文'],
    color='blue'
)
```
