```python
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = 'Arial Unicode MS'
```

```python
subjects = ['语文', '数学', '英语', '物理', '化学']
scores = [80, 90, 75, 95, 88]

# plt.barh(subjects, scores, color='blue', linestyle='-', label='线条1')
plt.bar(subjects, scores, color='orange', label='成绩',width=0.5)
plt.xlabel('科目')
plt.ylabel('成绩')
plt.title("小红2026成绩分布")
plt.legend(loc='upper left')

plt.grid(axis="y", alpha=0.5)

for a, b in zip(subjects, scores):
    plt.annotate(text=str(b), xy=(a, b+1))
plt.show()
```

```python
subjects = ['语文', '数学', '英语', '物理', '化学']
scores = [80, 90, 75, 95, 88]

plt.barh(subjects, scores, color='orange', label='成绩')
plt.xlabel('科目')
plt.ylabel('成绩')
plt.title("小红2026成绩分布")
plt.legend(loc='upper left')

plt.grid(axis="y", alpha=0.5)

for a, b in zip(subjects, scores):
    plt.annotate(text=str(b), xy=(b+1, a))

plt.show()
```
