```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 14, 8, 10]
plt.rcParams["font.family"] = 'Arial Unicode MS'
plt.figure(figsize=(10, 6))

plt.plot(x, y, color='blue', linestyle='-', label='线条1')
# 添加标题和坐标轴标签
plt.title('自定义样式的折线图')
plt.xlabel('X 轴')
plt.ylabel('Y 轴')
# 添加图例
plt.legend(loc='upper left')

#网格线
plt.grid(visible=True,  # 是否显示网格线

         which='major',  # 'major'(主网格), 'minor'(次网格), 'both'(两者)
         axis='both',  # 'x', 'y', 'both' 控制显示方向
         linestyle='--',  # 线型: '-', '--', ':', '-.'
         linewidth=0.5,  # 线条宽度
         color='gray',  # 线条颜色
         alpha=0.5  # 透明度 (0~1)
         )
# 添加annotation标注
for a, b in zip(x, y):
    plt.annotate(text=str(b), xy=(a, b))

plt.show()
```
