# 数组的基本使用

```
import numpy as np
```

```
arr0 = np.array(1)
print(arr0)
```

```
arr1 = np.array([1,3,4])
print(arr1)
print(f"数据的维度:{arr1.ndim}")
```

```
arr2 = np.array([[1,2,3], [4,5,6]])
print(arr2)
print(f"数据的维度:{arr2.ndim}")
```

# 数组的同质性 隐式转换

```
arr3 = np.array([1,'hello']) #1 隐式转换为了 字符串'1'
print(arr3)
```

# 数组的属性

```
arr4 = np.array([[1,2,3], [4,5,6]])
print(f"数组的维度:{arr4.ndim}")
print(f"数组的形状:{arr4.shape}")
print(f"数组的元素个数:{arr4.size}")
print(f"数组的数据类型:{arr4.dtype}")
print(f"数组的转置:{arr4.T}")
```
