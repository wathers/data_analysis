```python
import numpy as np
```

### 基础数据函数

```python
arr = np.array([1,2,3])
print(f"平方根,{np.sqrt(arr)}")
print(f"幂次计算,{np.power(arr,2)}")
print(f"指数计算,{np.exp(arr),np.exp2(arr)}")
print(f"对数计算,{np.log2(arr)}")
```

### 统计函数

```python
arr = np.random.randint(0,20,size=(2,8))
print(arr)

print(f"最大值,{np.max(arr)}")
print(f"最小值,{np.min(arr)}")
print(f"求和,{np.sum(arr)}")
print(f"平均值,{np.mean(arr)}")
print(f"中位数，{np.median(arr)}")
print(f"分位数(0.5),{np.quantile(arr,0.25,axis=1)}")
```
